```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
x_train.shape, y_train.shape , x_test.shape , y_test.shape
```

### By Linear Regression

```python
from sklearn.linear_model import LinearRegression
laptop_lr = LinearRegression()
laptop_lr.fit(x_train , y_train)
```

```python
print(laptop_lr.score(x_train,y_train))
print(laptop_lr.score(x_test,y_test))
```

```python
import joblib
joblib.dump(laptop_lr , "model.lb")
```

```python
laptop_1_price = laptop_lr.predict([[1,	0,	0,	1,	1,	0,	15.6,	1	,1]]) 
print(laptop_1_price)
```
